### Named Entity Recognition

```python
sentence="The Eiffel Tower was built from 1887 to 1889 by Gustave Eiffel, whose company specialized in building metal frameworks and structures."
```

```python
from pathlib import Path
import importlib.util

import nltk

if importlib.util.find_spec('numpy') is None:
    raise ModuleNotFoundError(
        "numpy is required for nltk.ne_chunk(). Install it in the active kernel with: pip install numpy"
    )

nltk_data_dir = Path.home() / 'nltk_data'
if str(nltk_data_dir) not in nltk.data.path:
    nltk.data.path.insert(0, str(nltk_data_dir))
```

```python
required_resources = {
    'tokenizers/punkt_tab': 'punkt_tab',
    'taggers/averaged_perceptron_tagger_eng': 'averaged_perceptron_tagger_eng',
    'chunkers/maxent_ne_chunker_tab': 'maxent_ne_chunker_tab',
    'corpora/words': 'words',
}

missing_packages = []
for resource_path, package_name in required_resources.items():
    try:
        nltk.data.find(resource_path)
    except LookupError:
        missing_packages.append(package_name)

if missing_packages:
    raise LookupError(
        "Missing NLTK data in the active environment: "
        + ", ".join(missing_packages)
        + ". Download them once with nltk.download(...) in an internet-enabled session."
    )

"NLTK resources are available."
```

```python
words=nltk.word_tokenize(sentence)
words
```

```python
tag_elements=nltk.pos_tag(words)
tag_elements
```

```python
named_entities = nltk.ne_chunk(tag_elements)
named_entities
```
